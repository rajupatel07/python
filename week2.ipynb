1. Calculate the total of `[25, 40, 15, 20]`.

```python
# Answers 1-3
# Write your answers here
num_lst=[25, 40, 15, 20]
print(sum(num_lst))
transactions=['paid', 'pending', 'paid']
for ind,val in enumerate(transactions):
    print(ind,val)
```

## Topic 6: Functions

1. Write `square(number)` that returns the number squared.
2. Write `full_name(first, last)` that returns a combined string.
3. Write `shipping_cost(amount, free_limit=100)` that returns 0 when the amount reaches the limit, otherwise 10.

```python
# Answers 1-3
# Write your answers here
def square(num1):
    return f'The square of number is {num1*num1}'
print(square(2))

def full_name(first,last):
    return f'My full name is {first} {last}.'
print(full_name('John','Doe'))

def shipping(amount,free_limit=100):
    print(f'0' if amount==free_limit else '10')
shipping(20)
```
