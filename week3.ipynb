## Topic 3: Nested Dictionaries

1. Sum the salaries in a nested `employees` dictionary.
2. Find the names of employees whose status is `Active`.
3. Add a `total_compensation` field to every employee using salary plus bonus.

```python
# Answers 1-3
# Write your answers here
employees = {
    'emp_1': {'name': 'John', 'salary': 50000, 'bonus': 5000, 'status': 'Active'},
    'emp_2': {'name': 'Alice', 'salary': 45000, 'bonus': 3000, 'status': 'Inactive'},
    'emp_3': {'name': 'Sam', 'salary': 52000, 'bonus': 4000, 'status': 'Active'},
}
total=[]
for emp in employees.values():
    total.append(emp['salary'])
print(sum(total))

Active_emp=[emp['name'] for emp in employees.values() if emp['status']=='Active']
print(f'Total Active Employee: {Active_emp}')
for key,value in employees.items():
    value['total_compensation']=value['salary']+value['bonus']
for key,value in employees.items():
    print(value['name'],value['total_compensation'])
```

## Topic 4: Conditional Statements

1. Classify an order as `small`, `medium`, or `large` using limits 100 and 300.

3. Handle a missing status with `.get()` and classify it as `unknown`.

```python
# Answers 1-3
# Write your answers here
def order_size(amount):
    if amount<100:
        return 'small'
    elif amount<300:
        return 'medium'
    else:
        return 'large'
print(order_size(500))

def valid_transaction(status,amount):
    if status=='OK' and amount>0:
        print(f'Transaction is approved of amount {amount}')
    else:
        print('Status must be OK and amount should be greater than 0!!!\nTry Again.')
valid_transaction('OK',100)

orders={
    'order_id':101,
    'amount':1000,
    'status':'Approved'
}

def status(log):
    if orders.get(log)==None:
        print('Unknown Entity')
    else:
        print(orders.get(log))
status('id')
```

## Topic 5: Loops

1. Calculate total revenue from a list of order dictionaries.
2. Count paid orders and collect the IDs of unpaid orders.
3. Validate records in a loop and collect IDs whose amount is missing or non-positive.

```python
# Answers 1-3
# Write your answers here
# 1. Calculate total revenue from a list of order dictionaries.
orders={
    "category":{
        "biscuits":{
            'marie':20,
            'goldie':10,
            'orieo':50,
            'creampie':30
        },
        'dairy':{
            'milk':30,
            'curd':40,
            'cheese':50,
            'lasi':20
        },
        'snacks':{
            'chips':10,
            'dorritos':5,
            'waffers':30
        }
    }
}

def func(opt):
    sum=0
    for key,value in orders['category'][f'{opt}'].items():
        sum+=value
    print(f'Total value of category {orders['category'][f'{opt}']} is {sum}')
func('snacks')
func('biscuits')
func('dairy')
```

```python
# 2. Count paid orders and collect the IDs of unpaid orders.

transaction={
    'order1':{
        'order_id':101,
        'item':'mobile',
        'pay_status':'paid'
    },
    'order2':{
        'order_id':102,
        'item':'laptop',
        'pay_status':'unpaid'
    },
    'order3':{
        'order_id':103,
        'item':'ipad',
        'pay_status':'paid'
    }
}
cnt=0
it=[]
for key,value in transaction.items():
    if value['pay_status']=='paid':
        cnt+=1
    else:
        it.append(value['order_id'])
print(f'Total paid orders are {cnt} and unpaid order Id\'s are {it}')
```
